```python
#Write a functionwhich generates a six digit/character random_user_id.
def random_user_id():
    user_id =[]
    for i in range(6):
        user_id.append(random.choice("0123456789abcdefghijklmnopqrstuvwxyz"))
    return "".join(user_id)
```

```python
def list_of_hexa_colors(n_colors):
    lst_hexa_colors= []
    chars= string.hexdigits[:16]
    for _ in range(n_colors):
        hex_color= "#" + "".join(random.choices(chars, k = 6))
        lst_hexa_colors.append(hex_color)
        
    return lst_hexa_colors
```

```python
def list_of_rgb_colors(n_rgb):
    lst_rgb_colors = []
    for _ in range(n_rgb):
        r, g, b = random.choices(range(256), k = 3)
        rgb_color= f"rgb({r}, {g}, {b})"
        lst_rgb_colors.append(rgb_color)
    
    return lst_rgb_colors
```

```python
def generate_colors(_type, n_colors):
    if _type == "hexa":
        return list_of_hexa_colors(n_colors)
    elif _type == "rgb":
        return list_of_rgb_colors(n_colors)
    else:
        return f"ERROR: {_type = } must be \"hexa\" or \"rgb\"" 
```

```python
   generate_colors('hexa', 3) # ['#a3e12f','#03ed55','#eb3d2b'] 
   generate_colors('hexa', 1) # ['#b334ef']
   generate_colors('rgb', 3)  # ['rgb(5, 55, 175','rgb(50, 105, 100','rgb(15, 26, 80'] 
   generate_colors('rgb', 1)  # ['rgb(33,79, 176)']
```

```python
def shuffle_list(lst):
    shuffled_lst = lst.copy()
    random.shuffle(shuffled_lst)
    
    return shuffled_lst
```
