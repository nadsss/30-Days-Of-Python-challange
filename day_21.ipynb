```python
class Statistics():
    def __init__(self,data):
        self.data = data.copy()
    def sum(self):
        sum = 0
        for i in self.data:
            sum += i
        return sum
    def count(self):
        count = 0
        for i in self.data:
            count += 1
        return count    
    def min(self):
        min = self.data[0]
        for i in self.data:
            if i < min:
                min = i
        return min
    def max(self):
        max = self.data[0]
        for i in self.data:
            if i > max:
                max = i
        return max
    def range(self):
        return self.max() - self.min()    
    def mean(self):
        return self.sum() / self.count()
    def median(self):
        lst = sorted(self.data)
        if self.count() % 2 == 0:
            return (lst[self.count()//2] + lst[(self.count()//2)-1]) //2
        else:
            return lst[self.count()//2]
    def mode(self):
        c = {}
        for num in self.data:
            if num not in c:
               c[num] = 1
            else:
               c[num] += 1
        c= sorted(c.items(), reverse = True, key = lambda x: x[1])
        m = [c[0][0]]
        for i in c:
            if i[1] == m[0]:
                m.append(i[0])
        return m
    def std(self):
        sum = 0
        for num in self.data:
            sum += (num - self.mean()) **2
        return (sum / self.count())**0.5
    def var(self) :
        sum = 0
        for num in self.data:
            sum += (num - self.mean()) **2
        return (sum / self.count())
    def freq_dist(self):
        freq = {}
        for x in self.data:
            freq[x] = freq.get(x, 0) + 1
        total = self.count()
        return sorted([(count / total * 100, value) for value, count in freq.items()], reverse=True) 
    def describe(self):
        return f"Count: {self.count()}\nSum: {self.sum()}\nMin: {self.min()}\nMax: {self.max()}\nRange: {self.range()}\nMean: {self.mean()}\nMedian: {self.median()}\nMode: {self.mode()}\nVariance: {self.var()}\nStandard Deviation: {self.std()}\nFrequency Dist: {self.freq_dist()}"
```

```python
ages = [31, 26, 34, 37, 27, 26, 32, 32, 26, 27, 27, 24, 32, 33, 27, 25, 26, 38, 37, 31, 34, 24, 33, 29, 26]
data = Statistics (ages)
```

```python
class PersonAccount():
    def __init__(self,firstname, lastname, incomes, expenses):
        self.firstname = firstname
        self.lastname= lastname
        self.incomes = incomes
        self.expenses = expenses
    def total_income(self):
        total = 0
        for i in self.incomes:
            total += i
        return total
    def total_expense(self):
        total = 0
        for i in self.expenses:
            total += i
        return total 
    def account_info(self):
         return f"firstname: {self.firstname}, lastname: {self.lastname}, incomes: {self.total_income()}, expenses: {self.total_expense()}"
    def add_income(self, income):
         self.incomes.append(income)
    def add_expense(self, expense):
         self.expenses.append(expense)
    def account_balance(self):
         return self.total_income() - self.total_expense()
```
